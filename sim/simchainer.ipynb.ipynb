```python
benergy = pd.read_csv('temp/Builder-Energy.txt', sep=' ', header=None)
e, h = benergy.iloc[:,0], benergy.iloc[:,1]
plt.figure(figsize=(6, 3))
plt.step(e, h)
gaussian = lambda x, a, b, c: a * np.exp(-(x - b)**2 / (2 * c**2))
fitrange = 50
popt, _ = curve_fit(gaussian, e[511-fitrange:511+fitrange], h[511-fitrange:511+fitrange])
plottingbounds = np.linspace(popt[1] - 3*popt[2], popt[1] + 3*popt[2], 1000)
plt.plot(plottingbounds, gaussian(plottingbounds, *popt), 'r-', label=f'Gaussian fit: a={popt[0]:.2f}, b={popt[1]:.2f}, c={popt[2]:.2f}')
print(popt[2] * 2.355 / popt[1])
plt.legend()
plt.show()
```
